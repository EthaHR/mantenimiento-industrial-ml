```python
import random
import pymysql

# Configuración de la conexión a MySQL
db_config = {
    'host': 'localhost',
    'user': 'root',
    'password': '',
    'database': 'mantenimiento_ml',
    'charset': 'utf8mb4'
}

# Parámetros de generación
TOTAL_REGISTROS = 30000
LOTE_TAMANO = 1000  # Inserción por lotes para mayor velocidad

print("Generando registros sintéticos...")

registros = []

for i in range(TOTAL_REGISTROS):
    horas_operacion = random.randint(100, 8000)
    dias_mantenimiento = random.randint(1, 180)
    
    # Simulación de variables con tendencia al desgaste
    temperatura = round(60.0 + (dias_mantenimiento * 0.1) + (horas_operacion * 0.001) + random.uniform(-5.0, 5.0), 2)
    vibracion = round(1.5 + (dias_mantenimiento * 0.005) + random.uniform(-0.5, 1.5), 2)
    presion = round(30.0 + random.uniform(-4.0, 4.0), 2)
    
    # Lógica condicional para definir si hubo falla (1) o normalidad (0)
    probabilidad_falla = 0.0
    if temperatura > 85.0:
        probabilidad_falla += 0.4
    if vibracion > 4.0:
        probabilidad_falla += 0.4
    if dias_mantenimiento > 120:
        probabilidad_falla += 0.2

    falla = 1 if random.random() < probabilidad_falla else 0

    registros.append((
        temperatura,
        vibracion,
        presion,
        horas_operacion,
        dias_mantenimiento,
        falla
    ))

print(f"Total de registros generados: {len(registros)}. Insertando en MySQL...")

# Conexión e inserción por lotes
try:
    conexion = pymysql.connect(**db_config)
    cursor = conexion.cursor()
    
    query = """
        INSERT INTO lecturas_maquinas 
        (temperatura, vibracion, presion, horas_operacion, dias_mantenimiento, falla)
        VALUES (%s, %s, %s, %s, %s, %s)
    """
    
    for start in range(0, TOTAL_REGISTROS, LOTE_TAMANO):
        lote = registros[start:start + LOTE_TAMANO]
        cursor.executemany(query, lote)
        conexion.commit()
        print(f"Insertados registros {start} a {min(start + LOTE_TAMANO, TOTAL_REGISTROS)}")

    print("¡Inserción masiva completada con éxito!")

except Exception as e:
    print(f"Error al conectar o insertar en la base de datos: {e}")

finally:
    if 'conexion' in locals() and conexion.open:
        cursor.close()
        conexion.close()
```
